# **Case Study – Student Performance Management System**

**Instructions**

Create two lists:
students
marks

Use range() to:
Access and print student names and marks using index positions

Use enumerate() to:
Display student index number and student name

Use zip() to:
Combine student names and marks together

Display all outputs using for loops

Use proper headings with print() statements

Print the final result in a clean format

```python
# for loops using range,enumerate and zip

student_names=["sudha","sudhiksha","madhu","lakshmi","sukanya"]
student_marks=[89,76,56,76,74]

print("----- Using range() -----")
#use range to print student names and marks
for i in range(len(student_names)):
  print(student_names[i], ":",student_marks[i])


print("------------\n ")
#use enumerate to display student index number and name
print("----- Using enumerate() -----")
for i,n in enumerate(student_names):
  print(i,"-",n)



print("------------\n ")
#use zip to combine student name and marks
print("----- Using zip() -----")
for name,mark in zip(student_names,student_marks):
  print(f"{name} has secured {mark} marks")

```
