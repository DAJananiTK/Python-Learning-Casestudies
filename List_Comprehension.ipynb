# **Case Study – String Analysis using List Comprehension**

**Instructions **

Create a string variable called text

Store any sentence or word inside the variable

Use list comprehension to:

Convert all characters into uppercase

Extract only vowels from the string

Extract only consonants from the string


Display:

Original string

Uppercase characters

Vowels list

Consonants list

String without spaces

Print the output in a clean format using print() statements

```python
#Create a string variable called text

# getting text as a user input

text=input("Enter the text: ")

#Use list comprehension
#Convert all characters into uppercase
uppercase_letters=[x.upper() for x in text]

#Extract only vowels from the string

vowels=[x for x in text if x.lower() in "aeiou"]

#Extract only consonants from the string
consonants=[x for x in text if x.isalpha() and x.lower() not in "aeiou"]


#String without spaces
without_spaces=[x for x in text if x!=" "]

#display all outputs
print(f"Original String : {text}")
print(f"Uppercase Characters :{uppercase_letters}")
print(f"Vowels List: {vowels}")
print(f"Consonants List: {consonants}")
print(f"String without spaces: {without_spaces}")
```

# **Case Study – Employee Salary Increment using List Comprehension**

**Instructions**

Create a list called salary

Store employee salaries inside the list

Use list comprehension to:
Increase each salary by 10%

Create another list to:
Display only salaries greater than 50000

Print:

Original salary list

Updated salary list

Filtered salary list


```python
#Create a list called salary
salary=[25000,30000,35000,40000,50000,56000,76000,89000]

#Use list comprehension to: Increase each salary by 10%
updated_salary=[round(x*1.10,2) for x in salary]


#Create another list to: Display only salaries greater than 50000
filtered_salary=[x for x in salary if x>50000]

#Print:
#Original salary list
#Updated salary list
#Filtered salary list

print(f"Original list        : {salary}")
print(f"Updated Salary List  : {updated_salary}")
print(f"Filtered Salary List : {filtered_salary}")



```
