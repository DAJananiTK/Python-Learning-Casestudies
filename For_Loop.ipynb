## **Case Study – Text Analyzer System**

**Instructions:**

1. Take a string input from the user using input().

2. Convert the text into lowercase using lower().

3. Use a for loop to iterate through each character in the string.

4. Count:
   • Total vowels  
   • Total consonants  
   • Total digits  
   • Total special characters  

5. Use:
   • isalpha() for alphabets  
   • isdigit() for numbers  

6. Display all counts using print()

```python
#for loop
user_input=input("Enter the input : ").lower()
vowels=0
consonants=0
digits=0
spl_characters=0
for i in user_input:
  if i in "aeiou":
    vowels+=1
  elif i.isalpha():
    consonants+=1
  elif i.isdigit():
    digits+=1
  else:
    spl_characters+=1
print(f"Number of vowels in the string : {vowels}")
print(f"Number of consonants in the string : {consonants}")
print(f"Number of digits in the string : {digits}")
print(f"Number of special characters in the string : {spl_characters}")
```

## **Case Study – Restricted Word Detection System**

**Instructions:**

1. Create a list of words containing normal and restricted words.

2. Use a for loop to read words one by one.

3. Check whether the current word is a restricted word using if condition.

4. If the restricted word is found:
   • Display "Restricted word detected"  
   • Stop the loop using break  

5. Otherwise display all safe words using print()

```python
#for loop with break
words=["hi","welcome","hello","blocked","allowed","granted"]
restricted_word="blocked"
for word in words:
  if word ==restricted_word:
    print("Restricted word Detected")
    break;

  print(f"Safe words : {word}")
```

## **Case Study – Employee Attendance Count System using continue**

**Instructions:**

1. Create a list named `attendance` with values:
   Present, Absent, Present, Leave, Present

2. Create count variables:

   • present  
   • absent  
   • leave  

3. Use a for loop to iterate through the attendance list.

4. Use continue statement to:
   • Skip remaining conditions after counting Present employees  

5. Count:
   • Total Present employees  
   • Total Absent employees  
   • Total Leave employees  

6. Display all counts using print()

```python
# for loop with continue

attendance=["Present","Absent","Present","Absent","Leave","Absent","Present","Absent","Leave"]
present=0
absent=0
leave=0
for status in attendance:
  if status=="Present":
    present+=1
    continue
  elif status=="Absent":
    absent+=1
  elif status =="Leave":
    leave+=1
print(f"Total Present Employees: {present}")
print(f"Total Absent Employees: {absent}")
print(f"Total Leave Employees: {leave}")
```
