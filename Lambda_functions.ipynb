# **Case Study – Employee Salary Analysis**

**Instructions**

Create a list called salary Store employee salaries inside the list

**Use Built-in Functions to:**

Find maximum salary

Find minimum salary

Find total salary

Find number of employees using len()

Use lambda with map() to:
Increase each salary by 10%

Use lambda with filter() to:
Display salaries greater than 50000

**Display**:

Original salary list

Updated salary list

High salary list

Maximum salary

Minimum salary

Total salary

```python
#Create a list called salary
#Store employee salaries inside the list

salary=[25000,30000,45000,57000,67000,89000,54000,64000]

#Find maximum salary
max_salary=max(salary)

#Find minimum salary
min_salary=min(salary)


#Find total salary
total_salary=sum(salary)

#Find number of employees using len()
number_of_employees=len(salary)

#Use lambda with map() to: Increase each salary by 10%
updated_salary=list(map(lambda x: round(x*1.10,2),salary))


#Use lambda with filter() to: Display salaries greater than 50000
high_salary=list(filter(lambda x: x>50000,salary))

#display all
print(f"Original salary list : {salary}")
print(f"Updated salary list : {updated_salary}")
print(f"High salary list : {high_salary}")
print(f"Maximum salary : {max_salary}")
print(f"Minimum salary : {min_salary}")
print(f"Total salary : {total_salary}")
print(f"Number of employees : {number_of_employees}")
```
