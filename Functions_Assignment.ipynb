```python
# 1. Find the largest of three numbers
def largest(a, b, c):
    return max(a, b, c)

print(f"Largest of three numbers : {largest(10, 25, 15)}")
```

```python
# 2. Factorial using recursion
def factorial(n):
    if n == 0 or n == 1:
        return 1
    return n * factorial(n - 1)

print(f"Factorial using recursion : {factorial(5)}")
```

```python
# 3. Check whether a string is a palindrome
def palindrome(text):
  t=text.lower()
  if t == text[::-1]:
    return ("Yes it is a palindrome")
  else:
    return ("It is not a palindrome")
print(palindrome("madam"))
```

```python
# 4. Return the sum of all elements in a list
def list_sum(numbers):
    return sum(numbers)

print(f"Sum of all elements in the list : {list_sum([10, 20, 30, 40])}")
```

```python
# 5. Find average using *args
def average(*args):
    avg =sum(args) / len(args)
    return avg

print(f"Average using *args: {average(10, 20, 30, 40)}")
```

```python
# 6. Count students who scored above 75
def count_above_75(marks):
    count = 0
    for mark in marks:
        if mark > 75:
            count += 1
    return count

print(f"Students count who scored above 75 : {count_above_75([80, 65, 90, 72, 88])}")
```

```python
# 7. Total sales greater than a given threshold
def total_sales_above_threshold(sales, threshold):
    total = 0
    for value in sales:
        if value > threshold:
            total += value
    return total

print(f"Total sales greater than the given threshold: {total_sales_above_threshold([1000, 5000, 3000, 8000], 4000)}")
```

```python
# 8. Count vowels in a string
def vowels_count(string):
  count=0
  lower=string.lower()
  for i in lower:
    if i in "aeiou":
      count+=1
  return count
print(f"vowels count in the given string :{vowels_count("BANana")}")
```

```python
# 9. Return only even numbers in a new list
def even_num(list):
  even_list=[]
  for i in list:
    if i%2 ==0:
      even_list.append(i)
  return even_list
print(f"List of even numbers : {even_num([4,5,6,7,8,9])}")
```

```python
# 10. Calculate Monthly Bonus for Employees
def calculate_bonus(employee_sales):
    bonus_list = []

    for sales in employee_sales:
        if sales > 50000:
            bonus = sales * 0.10
        elif sales >= 30000:
            bonus = sales * 0.05
        else:
            bonus = 0

        bonus_list.append(bonus)

    return bonus_list

employee_sales = [85000, 40000, 60000, 50000]
print(calculate_bonus(employee_sales))
```
