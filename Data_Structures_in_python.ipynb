Case Study

College Event Management System

Create a list called participants and perform add, insert, remove, and display operations.
Create a tuple called events and display the first event, last event, and total number of events.
Create a list called departments, add duplicate department names, convert the list into a set to remove duplicates, and perform add and remove operations.
Create a list called winners using tuples and convert the winners list into a frozenset.

```python
#Create a list called participants
participants=["guna","selva","kumar","ravi","nithin","jegan","guhan"]

#perform append, insert, remove, and display operations
participants.append("mahesh")
participants.insert(1,"naresh")
participants.remove("jegan")
print(f"Participants list : {participants}")
```

```python
#Create a tuple called events

events=("dance","singing","drawing","quiz","science expo","rangoli")

#display the first event, last event, and total number of events
print(f"First event : {events[0]}")
print(f"Last event : {events[-1]}")
print(f"Total number of events : {len(events)}")
```

```python
#Create a list called departments & add duplicate department names

departments = [
    "Human Resources",
    "Finance",
    "Marketing",
    "Sales",
    "Information Technology",
    "Operations",
    "Customer Support",
    "Research and Development",
    "Administration",
    "Finance",
    "Sales",
    "Operations"
]

#convert the list into a set to remove duplicates
dept_set=set(departments)


#perform add and remove operations
dept_set.add("Mechanical")
dept_set.remove("Operations")

print(f"Departments: {dept_set}")


```

```python
#Create a list called winners using tuples and convert the winners list into a frozenset.

winners_list = [
    ("Janani", "Gold"),
    ("Surya", "Silver"),
    ("Nila", "Bronze"),
    ("Megha","Silver")
]

# Convert the list into a frozenset
winner_set = frozenset(winners_list)


# Display
print("Winners List:", winners_list)
print("Frozen Set:", winner_set)

winners_list.append(("guhan","gold"))
print(winners_list)


```
